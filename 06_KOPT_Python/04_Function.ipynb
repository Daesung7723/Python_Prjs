화씨 계산식: 화씨 = 섭씨 * 1.8 + 32



```python
def convert_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

temp_in_fahrenheit = convert_to_fahrenheit(25)
print(f"25 degrees Celsius is {temp_in_fahrenheit} degrees Fahrenheit.")
```

### 미니 프로젝트: 텍스트 RPG (v2.0 상속 적용) ⚔️🛡️  

- 플레이어와 몬스터는 이름, 체력, 공격력을 가지고 공격을 한다는 공통점이 있습니다. 
- 이 공통점을 묶어 Character 라는 부모 클래스를 만들고, **Player**와 **Monster**가 Character의 능력을 물려받도록(상속) 만들어 보겠습니다.

#### 주요 학습 목표
- 클래스(Class): Character, Player, Monster 설계도를 만듭니다.
- 객체(Object): 실제 플레이어와 몬스터 객체를 생성합니다.
- 상속(Inheritance): Player와 Monster가 Character의 속성과 메소드를 물려받게 합니다.
- 모듈(Module): random 모듈로 전투에 무작위 요소를 추가합니다.

#### 게임의 핵심 기능
- Character (부모) 클래스 만들기
    - 모든 캐릭터의 공통 속성인 name, hp, power를 가집니다.
    - 공통 행동인 attack()과 show_status() 메소드를 가집니다.
- Player와 Monster (자식) 클래스 만들기
    - Character 클래스를 상속받습니다.
    - 상속만 받아도 attack 등 부모의 모든 능력을 바로 사용할 수 있습니다.
    - 여기에 더해, Player는 use_potion() 같은 자신만의 특별한 스킬을 추가할 수도 있습니다.
- 전투 시스템 구현 및 결과 출력

이전 계획과 동일하게 while 문과 random 모듈을 사용해 전투를 진행하고 결과를 출력합니다.



```python
import random as rd

class character:
    def __init__(self, name, hp, power):
        self.name = name
        self.hp = hp
        self.power = power

    def status(self):
        print(f"Name: {self.name}, HP: {self.hp}, Power: {self.power}")

    def attack(self):
        damage = rd.randint(1, self.power)
        print(f"{self.name} attacks and deals {damage} damage!")
        return damage

    def injured(self, damage):
        self.hp -= damage
        if self.hp <= 0:
            print(f"{self.name} has been defeated!")
        else:
            print(f"{self.name} has {self.hp} HP remaining.")
            print("-----")

class Player(character):
    def heal(self):
        heal_amount = rd.randint(10, 30)
        self.hp += heal_amount
        print(f"{self.name} heals for {heal_amount} HP! Current HP: {self.hp}")

class Monster(character):
    def roar(self):
        print(f"{self.name} lets out a terrifying roar!")

# Create instances
player = Player("Hero", 100, 20)
monster = Monster("Goblin", 80, 15)
# Display status
player.status()
monster.status()

# Simulate an attack
damage = player.attack()
monster.injured(damage)
damage = monster.attack()
player.injured(damage)

while player.hp > 0 and monster.hp > 0:
    print("                                         ")
    action = input("Choose action: 1. Attack 2. Heal\n")
    if action == '1':
        damage = player.attack()
        monster.injured(damage)
        if monster.hp <= 0:
            break
        damage = monster.attack()
        player.injured(damage)
    elif action == '2':
        player.heal()
        damage = monster.attack()
        player.injured(damage)
    else:
        print("Invalid action. Please choose again.")
```
